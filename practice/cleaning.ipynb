```python
import pandas as pd

df = pd.read_csv("data/processed/IT_Jobs_Cleaned.csv")

print(df.info())
```

```python
df[df.columns[-5:]].info()
```

```python
missing = df.isnull().sum()

missing_percent = (missing / len(df) * 100).round(2)

pd.DataFrame({
    "Missing": missing,
    "Missing %": missing_percent
}).sort_values("Missing %", ascending=False)
```

```python
df['Date Posted'] = pd.to_datetime(df['Date Posted'], errors='coerce')

df['Expected Last Date'] = pd.to_datetime(df['Expected Last Date'], errors='coerce')
```

```python
df.drop_duplicates(inplace=True)
```

```python
text_cols = df.select_dtypes(include='object').columns

for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
df.drop(columns=[
    'Apply Online if applicable',
    'Online Applicants',
    'Gender',
    'No. of Positions'
], inplace=True)
```

```python
df['Education'] = df['Education'].fillna('Not Mentioned')
```

```python
df['Newspaper'] = df['Newspaper'].fillna('Unknown')
```

```python
df['Area / Town'] = df['Area / Town'].fillna('Unknown')
```

```python
df['Job Type'] = df['Job Type'].str.title()
df['City'] = df['City'].str.title()
df['Organization'] = df['Organization'].str.title()
df['Job Industry'] = df['Job Industry'].str.title()
```

```python
df['Expected Salary'] = (
    df['Expected Salary']
      .str.replace(',', '', regex=False)
      .str.extract('(\d+)')
)
df['Expected Salary'] = pd.to_numeric(df['Expected Salary'])
```

```python
df['Posting Month'] = df['Date Posted'].dt.month_name()
```

```python
df['Posting Year'] = df['Date Posted'].dt.year
```

```python
df['Days Available'] = (
    df['Expected Last Date'] - df['Date Posted']
).dt.days
```
